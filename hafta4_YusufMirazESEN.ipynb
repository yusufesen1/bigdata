```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
import time, os, glob

spark = SparkSession.builder.appName("hafta4").master("local[4]").getOrCreate()
print("Spark surumu :", spark.version)

KOK = "work/gol" if os.path.isdir("work/gol") else "gol"
print("Yil klasoru :", len(glob.glob(KOK + "/yil=*")))

spark.conf.set("spark.sql.legacy.parquet.nanosAsLong", "true")
gol = spark.read.option("basePath", KOK).parquet(KOK + "/yil=*")
gol.createOrReplaceTempView("yorumlar")

spark.sql("SELECT COUNT(*) AS satir FROM yorumlar").show()
```
